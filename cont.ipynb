```python
contact = [   # contact list jisme har contact ek dictionary hai
    {'name': 'shivam', 'phone_num': ['1234567890', '9876543210']},   # shivam ke do number
    {'name': 'yash', 'phone_num': ['9890456312']},                   # yash ka ek number
    {'name': 'harsh', 'phone_num': ['9078652342', '9998887777']},    # harsh ke do number
    {'name': 'yashraj', 'phone_num': ['3334445555']},                # yashraj ka ek number
    {'name': 'satyajeet', 'phone_num': ['4445556666', '1231231234']},# satyajeet ke do number
    {'name': 'maloji', 'phone_num': ['5556667777']},                 # maloji ka ek number
    {'name': 'santosh', 'phone_num': ['6667778888', '7778889999']}   # santosh ke do number
]


def phone_number(pnumber):   # function jo phone number ko validate karega
    return len(pnumber) == 10 and pnumber.isdigit()   # valid number tab hoga jab 10 digit ho aur saare digits ho


def add_contact(contactlist, name, number):   # naya contact add karne ka function
    if phone_number(number) == False:   # pehle check karenge ki number valid hai ya nahi
        print("Invalid phone number")   # agar galat hai to message dikhayega
        return   # aur function yahi ruk jayega

    for c in contactlist:   # poore list me har contact check karenge
        if number in c["phone_num"]:   # agar number already list me hai
            print("Number already exists in contact list", c['name'])   # message dikhayega
            return   # aur ruk jayega

    for c in contactlist:   # ab check karenge ki naam already hai ya nahi
        if c["name"].lower() == name.lower():   # agar naam match hua (case-insensitive)
            c["phone_num"].append(number)   # existing contact me naya number add hoga
            print("Number added to existing contact",c['name'])   # message dikhayega
            return   # aur function yahi ruk jayega

    contactlist.append({"name": name, "phone_num": [number]})   # agar naam bhi new hai to naya contact add hoga
    print("New contact added")   # message dikhayega


def search_contact(contactlist, name):   # naam se contact search karne ka function
    for c in contactlist:   # poore contactlist check karenge
        if c["name"].lower() == name.lower():   # agar naam match hua
            print(f"{c['name']} : {c['phone_num']}")   # contact ka naam aur uske number print karega
            return   # aur function ruk jayega
    else:
            print("Contact not found")   # agar nahi mila to message dikhayega


def view_contact(contactlist):   # saare contacts dekhne ka function
    if len(contactlist)==0:   # agar list empty hai
        print("the contact list is empty")   # message dikhayega
    else:
        print(contactlist)   # warna poora list print karega


def update_contact(contactlist, name, new_number):   # contact update karne ka function
    if not phone_number(new_number):   # pehle new number valid hai ya nahi check karenge
        print("Invalid phone number")   # agar galat hai to message dikhayega
        return   # aur ruk jayega

    for c in contactlist:   # har contact check karenge
        if c["name"].lower() == name.lower():   # agar naam match hua
            c["phone_num"] = [new_number]   # purana list hatakar sirf new number rakhenge
            print("Number updated Sucessfully ")   # message dikhayega
            return   # aur ruk jayega
    else:
        print("Contact not found")   # agar naam nahi mila to message dikhayega


def delete_contact(contactlist, name):   # contact delete karne ka function
    for c in contactlist:   # har contact check karenge
        if c["name"].lower() == name.lower():   # agar naam match hua
            contactlist.remove(c)   # us contact ko delete karenge
            print("contact Deleted Sucessfully")   # message dikhayega
            return   # aur ruk jayega
    else:
        print("Contact not found")   # agar naam nahi mila to message dikhayega


while True:   # infinite loop (program baar-baar menu dikhata rahega jab tak exit nahi karte)
    print("--- Contact Book Menu ---")   # menu ka title
    print("1. Add a new contact")          # option 1
    print("2. View all contacts")          # option 2
    print("3. Search for a contact")       # option 3
    print("4. Update a contact")           # option 4
    print("5. Delete a contact")           # option 5
    print("6. Exit")                       # option 6

    try:
        choice = int(input("Enter your choice (1-6): "))   # user se choice input lete hai (number)
        if choice < 1 or choice > 6:   # agar 1 se 6 ke beech nahi hai
            print("Invalid choice! Enter number between 1 and 6.")   # message dikhayega
            continue   # aur loop fir se chalega
    except ValueError:   # agar user ne galti se number ke jagah text dal diya
        print("Invalid input! Please enter a number.")   # error message dikhayega
        continue   # aur loop fir se chalega

    if choice == 1:   # agar user ne 1 choose kiya
        name = input("Enter name: ")   # naam puchhega
        phone = input("Enter phone number: ")   # phone number puchhega
        add_contact(contact, name, phone)   # add_contact function call hoga

    elif choice == 2:   # agar user ne 2 choose kiya
        view_contact(contact)   # view_contact function call hoga

    elif choice == 3:   # agar user ne 3 choose kiya
        name = input("Enter name to search: ")   # naam puchhega
        search_contact(contact, name)   # search_contact function call hoga

    elif choice == 4:   # agar user ne 4 choose kiya
        name = input("Enter name to update: ")   # naam puchhega
        new_number = input("Enter new phone number: ")   # new number puchhega
        update_contact(contact, name, new_number)   # update_contact function call hoga

    elif choice == 5:   # agar user ne 5 choose kiya
        name = input("Enter name to delete: ")   # naam puchhega
        delete_contact(contact, name)   # delete_contact function call hoga

    elif choice == 6:   # agar user ne 6 choose kiya
        print("Exit")   # message dikhayega
           # loop tod dega aur program band ho jayega
    break
```
